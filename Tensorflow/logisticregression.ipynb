```python
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
y = np.random.rand(100)
```

```python
print(x)
```

```python
print(y)
```

```python
x_mean = np.mean(x)
y_mean = np.mean(y)
```

```python
m = np.sum((x-x_mean)*(y-y_mean)) / np.sum((x-x_mean)**2)
```

```python
b = y_mean - m*x_mean
```

```python
y_predict = m*x + b
```

```python
print(m)
```

```python
print(b)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y)
plt.plot(x ,y_predict)
```

```python
z = 1/(1+np.exp(-y_predict))
```

```python
plt.plot(x,z)
```
